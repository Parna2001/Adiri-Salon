```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_modal = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
embedding_modal.embed_query("what is AI?")
```

#### **STEP 4:Create and store Embeddings In Vector Store**


```python
from langchain_community.vectorstores import Chroma
vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_modal
)
```

```python
vectorstore.similarity_search("What is AI?", k= 3)
```

#### **Talk to LLM**


```python
vectorstore.similarity_search("What is AI?", k= 3)
```

```python
query = "What are the causes of earthquakes?"
retrieved_docs = vectorstore.similarity_search(query, k=3)
context = "\n\n".join(doc.page_content for doc in retrieved_docs)

response = llm.invoke(f"{query} You can answer using the following context: {context}")
print(response.content)
```

## **RAG IMPLEMENTATION WITH SALON CUSTOMER DATABASE (EXCEL)**

#### **STEP 1: Load the Excel file into Documents (one per customer)**

```python
import pandas as pd
from langchain_core.documents import Document

xlsx_path = r"C:\Users\srija\OneDrive\Desktop\VS.CODE\Salon Customers Database - Pradip Ray(Sheet1).xlsx"
df = pd.read_excel(xlsx_path)


def clean(v):
    """Return None for blanks, a plain string otherwise."""
    if pd.isna(v):
        return None
    if isinstance(v, float) and v.is_integer():
        v = int(v)
    return str(v).strip() or None


def row_to_text(r):
    name = " ".join(x for x in (clean(r.First_Name), clean(r.Middle_Name), clean(r.Last_Name)) if x)
    parts = [f"Customer: {name}."]
    mobiles = [m for m in (clean(r.Mobile_1), clean(r.Mobile_2)) if m]
    if mobiles:
        parts.append("Mobile: " + ", ".join(mobiles) + ".")
    if clean(r.Birth_Day) and clean(r.Birth_Month):
        parts.append(f"Birthday: {clean(r.Birth_Day)} {clean(r.Birth_Month).title()}.")
    if clean(r.Anniversary_Day) and clean(r.Anniversary_Month):
        parts.append(f"Anniversary: {clean(r.Anniversary_Day)} {clean(r.Anniversary_Month).title()}.")
    if clean(r.Notes):
        parts.append(f"Notes: {clean(r.Notes)}.")
    return " ".join(parts)


salon_docs = [
    Document(
        page_content=row_to_text(r),
        metadata={
            "source": "Salon Customers Database",
            "row": i + 2,  # Excel row number (header is row 1)
            "birth_month": (clean(r.Birth_Month) or "").title(),
            "anniversary_month": (clean(r.Anniversary_Month) or "").title(),
        },
    )
    for i, r in enumerate(df.itertuples(index=False))
]
print(len(salon_docs), "customers loaded")
salon_docs[:3]
```

#### **STEP 2: Create and store Embeddings in a Vector Store**

Each customer is already one short record, so no chunking is needed.

```python
salon_store = Chroma.from_documents(
    documents=salon_docs,
    embedding=embedding_modal,
    collection_name="salon_customers",
)
```

#### **STEP 3: Talk to LLM**

```python
def ask_salon(query, k=10):
    retrieved = salon_store.similarity_search(query, k=k)
    context = "\n".join(d.page_content for d in retrieved)
    response = llm.invoke(
        "You are answering questions about a salon's customer database. "
        "Use only the customer records below. If the answer isn't in them, say so.\n\n"
        f"Records:\n{context}\n\nQuestion: {query}"
    )
    return response.content


print(ask_salon("Which customers have a birthday in September?"))
```
